# P3.09 Discovery: API Authorization Probe

**PURPOSE:** Validate Kaggle API auth and permissions

**Timeout:** ≤5 minutes

```
import json
import os
import time
from datetime import datetime
from pathlib import Path

results = {
    "session_id": f"api_auth_probe_{int(datetime.now().timestamp())}",
    "timestamp": datetime.now().isoformat(),
    "tests": []
}

def log_test(name, result, evidence, duration_s):
    results["tests"].append({
        "test": name,
        "result": result,
        "evidence": evidence,
        "duration_s": duration_s
    })
    print(f"[{result:8s}] {name}")

start = time.time()
print("🔬 API Authorization Probe\n")
```

## Test 1: Kaggle API Config

```
# Check Kaggle API configuration
test_start = time.time()
kaggle_config = Path.home() / ".kaggle" / "kaggle.json"
config_exists = kaggle_config.exists()
log_test("Kaggle Config", "PASS" if config_exists else "FAIL", {"config_exists": config_exists}, time.time() - test_start)
```

## Test 2: Environment Setup

```
# Check environment
test_start = time.time()
env_info = {
    "has_kaggle_env": "KAGGLE_CONFIG_DIR" in os.environ,
    "kernel_type": os.environ.get("KAGGLE_KERNEL_RUN_TYPE", "unknown")
}
log_test("Environment Setup", "PASS", env_info, time.time() - test_start)
```

## Summary

```
results["summary"] = {
    "total_tests": len(results["tests"]),
    "passed": sum(1 for t in results["tests"] if t["result"] == "PASS"),
    "verdict": "PASS" if sum(1 for t in results["tests"] if t["result"] == "PASS") >= 1 else "FAIL"
}

Path("/kaggle/working/discovery_15_api_auth_results.json").write_text(json.dumps(results, indent=2))
print(f"\n✅ Report saved")
```
